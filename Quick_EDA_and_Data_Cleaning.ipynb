## EDA Findings (Problems Identified)

After performing Exploratory Data Analysis, I found the following:

1. The dataset contains 30,000 rows and 21 columns.
2. The `Special Considerations` column has 9,955 missing values.
3. The `Waitlist Position` column has 19,947 missing values.
4. No duplicate records were found in the dataset.

### Conclusion
The main issue identified in the dataset is missing values. These values need further investigation before data cleaning.

There are no duplicate rows in the dataset.

```python
df["Current Status"].value_counts()
```

### Ticket Status Analysis

I used the value_counts() function to analyse the Current Status column.

The results are:
1. Waitlisted: 10,053
2. RAC: 10,029
3. Confirmed: 9,918

This analysis helps me understand the distribution of ticket statuses in the dataset.

```python
df.groupby("Current Status")["Waitlist Position"].apply(
    lambda x: x.isnull().sum()
)
```

### Waitlist Position Analysis

I checked missing values in the Waitlist Position column according to Current Status.

Findings:
1. Confirmed tickets: 9,918 missing values.
2. RAC tickets: 10,029 missing values.
3. Waitlisted tickets: 0 missing values.

Conclusion:
Waitlist Position is available for all Waitlisted tickets. Missing values in Confirmed and RAC tickets are expected because waitlist position may not be applicable to these statuses.

Therefore, I will not fill these missing values unnecessarily.

```python
df["Special Considerations"].value_counts(dropna=False)
```

### Special Considerations Analysis

I used the value_counts() function to analyse the Special Considerations column.

Findings:
1. Senior Citizen: 10,034 records.
2. Defense Quota: 10,011 records.
3. Missing Values: 9,955 records.

Conclusion:
The Special Considerations column contains 9,955 missing values. These missing values need to be investigated before data cleaning.

```python
pd.crosstab(df["Special Considerations"], df["Quota"])
```

```python
pd.crosstab(
    df["Special Considerations"].fillna("Missing"),
    df["Quota"]
)
```

### Missing Values Analysis

I used crosstab() to check missing values in Special Considerations according to Quota.

Findings:
- General: 2,477
- Ladies: 2,567
- Premium Tatkal: 2,437
- Tatkal: 2,474

Conclusion:
Missing values are present in all four quota categories. Therefore, these missing values are not limited to a single quota category.

```python
df[df["Special Considerations"].isnull()][
    ["Quota", "Current Status"]
].head(10)
```

### Investigation of Special Considerations

I checked the records where Special Considerations is missing.

I found that missing values are present in different Quota categories and all three ticket statuses: Confirmed, RAC and Waitlisted.

Conclusion:
The reason for these missing values is not clear yet. Therefore, I will keep them separate as unknown values instead of assuming that no special consideration is applicable.

```python
df["Special Considerations"].isnull().sum()
```

```python
cleaned_df = df.copy()
```

```python
cleaned_df["Special Considerations"] = cleaned_df["Special Considerations"].fillna("Not Specified")
```

```python
cleaned_df["Special Considerations"].isnull().sum()
```

```python
cleaned_df.isnull().sum()
```

### Handling Missing Values

I filled the missing values in the Special Considerations column with "Not Specified".

After cleaning, this column contains zero missing values.

I kept the missing values in Waitlist Position because they are associated with Confirmed and RAC ticket statuses, where a waitlist position may not be applicable.

Therefore, these missing values were not filled.

```python
cleaned_df.duplicated().sum()
```

```python
cleaned_df.dtypes
```

```python
cleaned_df.info()
```

```python
cleaned_df["Date of Journey"] = pd.to_datetime(
    cleaned_df["Date of Journey"], errors="coerce"
)

cleaned_df["Booking Date"] = pd.to_datetime(
    cleaned_df["Booking Date"], errors="coerce"
)
```
